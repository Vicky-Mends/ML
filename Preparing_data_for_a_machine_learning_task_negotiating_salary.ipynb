```python
X= X.copy()
```

```python
X["degree_required"].value_counts()
```

```python
#ordinal encoding
```

```python
map_dict = {"none": 0,"bachelors":1,"masters":2,"phd":3}
X["degree_required"]=X["degree_required"].map(map_dict)
```

```python
X["company_type"].value_counts()
```

```python
#OHE
```

```python

X = pd.get_dummies(X,columns= ["company_type"],dtype= np.int32)
```

```python

X
```

```python
X
```

```python
X["metro"].value_counts()
```

```python
X
```

```python
#grouping rare metro areas into "Other" so the model can learn better and handle unfamiliar locations.
```

```python
frequent_metros = X["metro"].value_counts()
frequent_metros = frequent_metros[frequent_metros >= 100].index

X.loc[~X["metro"].isin(frequent_metros), "metro"] = "Other"

X = pd.get_dummies(X, columns=["metro"], dtype=np.int32)
```

```python
X
```

```python
#working on the last column
```

```python
vec= CountVectorizer()
vec.fit(X["job_text"])
```

```python
vec.get_feature_names_out()
```

```python
X_txt= vec.transform(X["job_text"])
```

```python
X_txt
```

```python
X_txt.toarray()
```

```python
X["job_text"].iloc[0]
```

```python
X
```

```python
#adding X_txt back into X
```

```python
col_names = ["job_text_" + n for n in vec.get_feature_names_out() ]
X_txt_df = pd.DataFrame(X_txt.toarray(), columns = col_names)
```

```python
X_txt_df
```

```python
X= pd.concat([X_txt_df,X],axis =1)
```
